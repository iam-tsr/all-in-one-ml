```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('../../datasets/titanic-dataset.csv')

# Simple Preprocessing
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Fare'] = df['Fare'].fillna(df['Fare'].median())

# Select basic features
features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']
X = df[features].values
y = df['Survived'].values.reshape(-1, 1)

# Normalization (Standardization) - Crucial for Neural Networks
X = (X - X.mean(axis=0)) / X.std(axis=0)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(z):
    return sigmoid(z) * (1 - sigmoid(z))

# Hyperparameters
input_size = 6    # Number of features
hidden_size = 8   # Neurons in hidden layer
output_size = 1   # Binary survival (0 or 1)
learning_rate = 0.1
epochs = 1000

# Initialize weights and biases
np.random.seed(42)
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Training Loop
for epoch in range(epochs):
    # --- FORWARD PASS ---
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2) # Output layer activation (predictions)
    
    # --- BACKWARD PASS ---
    # Error at output
    error = a2 - y
    d_z2 = error * sigmoid_derivative(z2)
    
    # Error at hidden layer
    d_z1 = np.dot(d_z2, W2.T) * sigmoid_derivative(z1)
    
    # --- UPDATE WEIGHTS (Gradient Descent) ---
    W2 -= learning_rate * np.dot(a1.T, d_z2) / len(X)
    b2 -= learning_rate * np.sum(d_z2, axis=0) / len(X)
    W1 -= learning_rate * np.dot(X.T, d_z1) / len(X)
    b1 -= learning_rate * np.sum(d_z1, axis=0) / len(X)

    if epoch % 200 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss:.4f}")

# Final Prediction (Threshold at 0.5)
predictions = (a2 > 0.5).astype(int)
accuracy = np.mean(predictions == y)
print(f"\nFinal Training Accuracy: {accuracy * 100:.2f}%")
```

## Feedforward Neural Network (FNN)

A FFN is the simplest type of artificial neural network. In this architecture, information moves in only one direction—forward—from the input nodes, through any hidden layers, to the output nodes. There are no cycles or loops in the network.

A standard FNN consists of three main components:

- **Input Layer:** Receives the raw data.

- **Hidden Layers:** One or more layers that perform intermediate computations to extract features.

- **Output Layer:** Produces the final prediction or classification.

#### **1. The Mathematical Model**

To understand how data flows, we look at a single layer. Each neuron in a layer calculates a weighted sum of its inputs and applies an activation function.

**Step 1: Linear Combination**

For a layer $l$, the input vector $a^{(l−1)}$ is transformed into a pre-activation vector $z^{(l)}$ using a weight matrix $W^{(l)}$ and a bias vector $b^{(l)}$:

$$z^{(l)}=W^{(l)}a^{(l−1)}+b^{(l)}$$

**Step 2: Activation Function**

To introduce non-linearity (allowing the network to learn complex patterns), we apply an activation function $σ$ (such as ReLU or Sigmoid) to $z^{(l)}$:

$$a^{(l)}=σ(z^{(l)})$$

#### **3. The Forward Pass**

The process of calculating the output of the entire network is a chain of these operations. If we have a network with $L$ layers, the final output $\hat{y}$​ is:

$$\hat{y} = a^{(L)} = \sigma^{(L)} \left(W^{(L)} \left(\cdots \sigma^{(1)} \left( W^{(1)} x + b^{(1)} \right)\cdots\right)+ b^{(L)}\right)$$

#### **4. Learning via Backpropagation**

While the feedforward part describes how data moves to generate an output, the network learns by comparing $\hat{y}$​ to the true value $y$ using a Loss Function $C$.

The weights are updated using Gradient Descent to minimize this loss:

$$W_{\text{new}} = W_{\text{old}} - \eta \frac{\partial C}{\partial W}$$

where $η$ is the learning rate.

```python
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('../../../datasets/titanic-dataset.csv')
df.head()
```

```python
features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']
X = df[features].copy()
y = df['Survived']
```

```python
# Encode 'Sex' to numbers (0 and 1)
X['Sex'] = LabelEncoder().fit_transform(X['Sex'])
```

```python
X.isnull().sum()
```

```python
# Impute missing ages with the median
X['Age'] = SimpleImputer(strategy='median').fit_transform(X[['Age']])
```

```python
# We use 5 folds, meaning 80% train / 20% test in each iteration
model = RandomForestClassifier(n_estimators=100, random_state=42)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
# Execute Cross-Validation

# 'cv' can take an integer or a cross-validation generator (like skf)
cv_results = cross_val_score(model, X, y, cv=skf, scoring='accuracy')

print(f"Accuracy per fold: {cv_results}")
print(f"Overall Model Accuracy: {cv_results.mean():.2%}")
print(f"Standard Deviation: {cv_results.std():.4f}")
```

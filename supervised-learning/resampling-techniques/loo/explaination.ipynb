## Leave-One-Out Cross-Validation (LOOCV)

LOOCV is a specific type of k-fold cross-validation where the number of folds $k$ is equal to the $n$ number of data points in the dataset.

In this approach, the model is trained $n$ times. For each iteration, a single observation is held out as the test set, while the remaining $n−1$ observations form the training set. This process is repeated until every single data point has been used exactly once as the test data.

Suppose we have a dataset $D$ with n observed data points:

$$D = \{(x_1, y_1), (x_2, y_2), \ldots, (x_n, y_n)\}$$

For each $i∈\{1,…,n\}$:

- Training Set: $D_{(−i)​}=D\{(x_i​,y_i​)\}$ (all points except the $i$-th).

- Model Fitting: Train the model $\hat{f}_{(-i)}$​ using $D_{(−i)}$​.

- Prediction: Calculate the prediction $\hat{y}_{​i}​=\hat{f}_{​(−i)}​(x_i​)$.

- Error Calculation: Compute the error for that specific point, usually using Mean Squared Error (MSE) for regression:

$$MSE_i = (y_i - \hat{y}_i)^2$$

**Final Error Estimation**

The total LOOCV estimate for the test error is the average of these n individual test errors:

$$CV_{(n)} = \frac{1}{n} \sum_{i=1}^{n} MSE_i$$

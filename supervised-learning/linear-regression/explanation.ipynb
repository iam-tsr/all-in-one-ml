## Linear Regression Algorithm

Linear regression is a fundamental statistical and machine learning method used to model the relationship between a dependent variable (target) and one or more independent variables (predictors). The goal is to find the linear equation that best describes the data points.

Imagine you have a scatter plot of data. Linear regression attempts to draw a straight line through those points in a way that minimizes the distance between the line and the actual data.


```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 30, 50)
y = 2 * x + 1

plt.plot(x, y, label='Linear Fit', color='blue')
plt.scatter(x, y + np.random.normal(0, 5, size=x.shape), label='Data Points', color='red')
plt.title('Simple Linear Regression')
plt.legend()
plt.grid()
plt.show()
```

In its simplest form (Simple Linear Regression), the relationship is modeled by the equation:

$$y = β_0​+β_1​x$$

Where:
- $y$: The predicted value (Dependent variable).

- $x$: The input feature (Independent variable).

- $β_0$​: The y-intercept, representing the value of $y$ when $x$=0.

- $β_1$​: The slope, representing the change in $y$ for every one-unit change in $x$.

**How it learns?**

To find the "best" line, we need to minimize the errors. The most common method is Ordinary Least Squares (OLS). It works by minimizing the Sum of Squared Errors (SSE):

$$SSE = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

We take the distance between each actual data point ($y_i​$) and the predicted point on the line ($y^{​i}$​), square that distance to remove negative values, and add them all up. The line that results in the lowest possible sum is our winner.

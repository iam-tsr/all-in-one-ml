## k-Nearest Neighbors (k-NN)

It is a simple, versatile, and "lazy" learning algorithm used for both classification and regression. It operates on a remarkably intuitive principle: similar things exist in close proximity.

Instead of building a complex model, KNN stores the entire training dataset and makes predictions based on the similarity between a new data point and its closest neighbors.

**How the Algorithm Works**

- Choose the number k: This is the number of nearest neighbors to check (e.g., k=3 or k=5).

- Calculate Distance: Measure the distance between the new point and all points in the training set.

- Identify Neighbors: Select the k points with the smallest distance to the new point.

- Vote or Average:

    - Classification: The new point is assigned the class most common among its k neighbors.
    
    - Regression: The new point is assigned the average value of its k neighbors.

To determine which neighbors are "nearest," KNN uses distance metrics. The most common is the **Euclidean Distance**.

**Euclidean Distance**

In a 2D space with two points $P1​(x_1​,y_1​)$ and $P2​(x_2​,y_2​)$, the distance $d$ is:

$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

The value of k significantly impacts the model's performance:

- Small k (e.g., k=1): The model is sensitive to noise and outliers (High variance/Overfitting).

- Large k: The model becomes smoother but may ignore local patterns (High bias/Underfitting).

- Pro-tip: For binary classification, it is best to choose an odd number for k to avoid "tie" votes between classes.

## Linear Discriminant Analysis (LDA)

It is a technique used in machine learning and statistics for dimensionality reduction and classification.

While PCA (Principal Component Analysis) focuses on finding the direction of maximum variance, LDA focuses on finding a feature space that maximizes the separability between classes. Think of it as trying to find the best camera angle to take a photo of two different groups of people so they appear as far apart as possible.

LDA aims to project data onto a lower-dimensional space while maintaining two goals:

- **Maximize the distance** between the means of different classes.

- **Minimize the spread** (variance) within each individual class.

To find the optimal projection, we use two "scatter" matrices. Let’s say we have data with d features and C classes.

1. **Within-Class Scatter Matrix $(S_W​)$:** This measures how spread out the data points are within their own groups. We want this to be small.

$$S_W = \sum_{i=1}^{C} \sum_{x \in \mathrm{Class}_i} (x - m_i)(x - m_i)^T$$
Where $m_i$​ is the mean vector of class $i$.

2. **Between-Class Scatter Matrix $(S_B​)$:** This measures the distance between the means of the different classes. We want this to be large.

$$S_B = \sum_{i=1}^{C} N_i (m_i - m)(m_i - m)^T$$

Where $m$ is the overall mean of all data and $N_i​$ is the number of samples in class $i$.

3. **The Optimization Goal:** We seek a projection vector w that maximizes the Fisher’s Criterion:

$$J(w) = \frac{w^T S_B w}{w^T S_W w}$$

By solving this as an eigenvalue problem, we find that the optimal $w$ corresponds to the eigenvectors of the matrix:

$$S_W^{-1}S_B$$

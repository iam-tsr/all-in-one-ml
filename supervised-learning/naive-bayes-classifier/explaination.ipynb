## Naive Bayes Classifier

The Naive Bayes Classifier is a probabilistic machine learning model used for classification tasks. It is based on the Bayes’ Theorem and is called "naive" because it makes the simplifying assumption that all input features are independent of each other given the class label.

The core of the classifier is Bayes' Theorem, which calculates the posterior probability of a class $C$ given a set of features $x_1​,x_2​,...,x_n​$.

$$P(C \mid x_1, x_2, \ldots, x_n) =\frac{P(x_1, x_2, \ldots, x_n \mid C)\, P(C)}{P(x_1, x_2, \ldots, x_n)}$$

Where:
- $P(C∣x)$ (Posterior Probability): The probability that a sample belongs to class $C$ given the features.

- $P(C)$ (Prior Probability): The overall probability of class $C$ appearing in the dataset.

- $P(x∣C)$ (Likelihood): The probability of the features occurring given that we know the class is $C$.

- $P(x)$ (Evidence): The total probability of the features occurring across all classes (used as a normalization constant).

**The "Naive" Assumption**

In a real-world scenario, features usually depend on each other. However, Naive Bayes assumes that:

$$P(x_1, x_2, \ldots, x_n \mid C)=P(x_1 \mid C)\cdot P(x_2 \mid C)\cdots P(x_n \mid C)$$

By substituting this into the original equation, we get the simplified classification formula:

$$P(C \mid x_1, \ldots, x_n) \propto P(C) \prod_{i=1}^{n} P(x_i \mid C)$$

The model predicts the class $\hat{y}$​ that maximizes this value:

$$\hat{y} = \argmax{_y}\ \left( P(C_y) \prod_{i=1}^{n} P(x_i \mid C_y) \right)$$

**Core Use Cases**

Primarily used for classification tasks, such as:
- Spam Filtering: Identifying junk mail by calculating the probability of words like "free" or "winner" appearing in spam vs. legitimate emails.

- Sentiment Analysis: Categorizing text—such as social media posts or product reviews—as positive, negative, or neutral.

- Document Categorization: Automatically sorting news articles, blog posts, or emails into topics like sports, technology, or politics.

- Recommendation Systems: Predicting whether a user will like a product based on their past behavior and demographic data.

- Medical Diagnosis: Assisting doctors in predicting the likelihood of a disease based on patient symptoms.

- Real-Time Prediction: Because the algorithm is computationally efficient and fast, it is often used for instant classification in live applications.

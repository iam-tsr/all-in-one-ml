## Support Vector Machine (SVM)

Support Vector Machine (SVM) is a powerful supervised learning algorithm used for both classification and regression, though it is most famous for binary classification.

The core idea is to find the optimal hyperplane that separates data points of different classes with the maximum possible margin.

In a $d$-dimensional space, a hyperplane is a flat affine subspace of dimension $d−1$. Mathematically, it is defined by the equation:

$$w^{T}x+b=0$$

Where:
- w is the weight vector (normal to the hyperplane).

- b is the bias (offset from the origin).

- x is the input feature vector.

**The Decision Rule**

For a binary classification problem with labels $y_i​∈$ {$−1,1$}, the SVM seeks a hyperplane such that:

- $w^{T}x_i​+b≥1$ if $y_i​=1$

- $w^{T}x_i​+b≤−1$ if $y_i​=−1$

These can be combined into a single constraint:

$$y_i​(w^{T}x_i​+b)≥1$$

**Maximizing the Margin**

The "Support Vectors" are the data points closest to the hyperplane, lying exactly on the boundaries wTx+b=±1. The distance between these two boundaries (the margin) is calculated as:

$$\text{Margin} = \frac{2}{\|w\|}$$

**The Kernel Trick**

When data is not linearly separable in its original space, SVM uses a Kernel Function $K(x_i​,x_j​)$. This implicitly maps the data into a higher-dimensional space where a linear separation becomes possible without actually performing the expensive transformation coordinates.

Common kernels include:

- **Linear:** $K(x,z)=x^{T}z$

- **Polynomial:** $K(x,z)=(1+x^{T}z)^{d}$
